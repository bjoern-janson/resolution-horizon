# Resolution Horizon — Live Demo
Click the play button on the code block below to run the simulation and plot the horizon curves!

```python
import numpy as np
import matplotlib.pyplot as plt
from instrument import StructuralHorizonEstimator

# 1. Initialize the instrument
instrument = StructuralHorizonEstimator(N=100000, sigma=0.03)
metrics = instrument.evaluate_horizon(max_k=7, lambda_N=1.0)

# 2. Extract curves
k_axis = metrics['k_axis']
eta = metrics['eta']
k_star = metrics['k_star']

# 3. Plot the result
plt.figure(figsize=(8, 4))
plt.plot(k_axis, eta, 'o-', color='indigo', label='Efficiency ($\\eta$)')
plt.axvline(x=k_star, color='crimson', linestyle='--', label=f'Horizon (k*={k_star})')
plt.title('Structural Resolution Horizon Transition')
plt.xlabel('Algebraic Depth (k)')
plt.ylabel('Information Efficiency')
plt.legend()
plt.grid(True)
plt.show()
```
